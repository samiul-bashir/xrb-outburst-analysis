# 03 — FRED fitting: pure, tophat, and asymmetric variants

Report sections 5.1 and 5.2. Fits each detected outburst peak with a Fast-Rise
Exponential-Decay profile. Three model variants are available (see
`docs/algorithms.md` §4); this notebook shows all three side by side on one source
so you can compare chi2_red before choosing which to run at scale.

```python
import sys
sys.path.insert(0, '../src')
import matplotlib.pyplot as plt

from xrb_pipeline.detection.simple_detector import run_detection_all, DEFAULT_CONFIG
from xrb_pipeline.fitting import pure_fred, fred_tophat, asymmetric_burst_profile as abp
```

```python
SOURCES = ["J0115+634"]   # 4U 0115+634, a Be/X-ray binary with many outbursts

outburst_results = run_detection_all(SOURCES, DEFAULT_CONFIG)
```

## Pure FRED

```python
name = SOURCES[0]
fred_fits = pure_fred.fit_fred_all(name, outburst_results, DEFAULT_CONFIG)
for fit in fred_fits[:3]:
    pure_fred.plot_fit(fit, title=name)
```

## FRED + Tophat

Usually a dramatically better chi2_red on outbursts with a sustained peak.

```python
tophat_fits = fred_tophat.fit_fred_tophat_all(name, outburst_results, DEFAULT_CONFIG)
for fit in tophat_fits[:3]:
    fred_tophat.plot_fit(fit, title=name)
```

## Asymmetric burst profile (alternative model)

Independent rise/decay amplitudes plus a free background. Not used in the final
report results, but a useful alternative when rise and decay genuinely differ in
characteristic scale.

```python
asym_fits = abp.fit_burst_profile_all(name, outburst_results, DEFAULT_CONFIG)
print(f"{len(asym_fits)} asymmetric fits")
```

## Save fitted FRED+tophat windows to disk

```python
all_fits = fred_tophat.fit_fred_tophat_for_all_sources(outburst_results, DEFAULT_CONFIG)
summary_df = fred_tophat.save_fits_to_disk(all_fits, outdir='../data/fits/outburst_fits')
summary_df.head()
```
