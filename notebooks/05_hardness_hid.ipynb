## One outburst at a time

`inspect_outburst_hid` fetches MAXI (multiband) and ZTF, restricts to the outburst

pair (g-r, falling back to r-i) exists in that window.

```python
MAXI_ID = "J1911+005"      # Aql X-1
ZTF_ID = "ZTF18accedeu"
X1, X2 = 57580.0, 57750.0   # outburst window, from your detection/inspection step

hid.inspect_outburst_hid(MAXI_ID, ZTF_ID, X1, X2, source_name="Aql X-1", savedir='../data/hid')
```
